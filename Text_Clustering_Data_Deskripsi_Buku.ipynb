# Chapter 4 - Text Pre-processing Data Books to Scrape

```python
import requests
from bs4 import BeautifulSoup
from urllib.parse import urljoin

import pandas as pd
import re
import string
import nltk
import matplotlib.pyplot as plt

from collections import Counter
from wordcloud import WordCloud

from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
```

```python
nltk.download('stopwords')
```

```python
data = []
```

```python
base_url = "https://books.toscrape.com/"

data = []

for page in range(1, 51):
    if page == 1:
        url = base_url
    else:
        url = f"{base_url}catalogue/page-{page}.html"

    response = requests.get(url)
    soup = BeautifulSoup(response.text, "html.parser")
    books = soup.find_all("article", class_="product_pod")

    for book in books:
        title = book.h3.a.get("title", "")

        detail_url = urljoin(url, book.h3.a.get("href", ""))

        detail_response = requests.get(detail_url)
        detail_soup = BeautifulSoup(detail_response.text, "html.parser")

        description = ""

        description_tag = detail_soup.find("div", id="product_description")

        if description_tag:
            description_p = description_tag.find_next_sibling("p")

            if description_p:
                description = description_p.get_text(" ", strip=True)

        data.append([title, description])

df = pd.DataFrame(data, columns=["Title", "Description"])

print("Jumlah data:", len(df))
```

```python
def remove_repeated_text(text):
    text = str(text).strip()

    words = text.split()

    # Cari pola pengulangan dengan panjang minimal 10 kata
    for n in range(min(100, len(words) // 2), 9, -1):
        first_part = words[:n]

        for start in range(n, len(words) - n + 1):
            if words[start:start+n] == first_part:
                # Hapus bagian yang berulang
                words = words[:start] + words[start+n:]
                return " ".join(words)

    return text

df["Description"] = df["Description"].apply(remove_repeated_text)
```

## 1. Remove HTML Tags

Menghapus tag HTML yang masih terdapat pada teks jika ada.

```python
def remove_html(text):
    return re.sub(r'<.*?>', '', str(text))

df["Text_Clean"] = df["Text"].apply(remove_html)
df[["Text", "Text_Clean"]].head(100)
```

## 2. Lower Casing

Mengubah seluruh huruf menjadi huruf kecil agar kata yang sama dengan perbedaan kapitalisasi dianggap sebagai kata yang sama.

```python
df["Text_Clean"] = df["Text_Clean"].str.lower()
df[["Text", "Text_Clean"]].head(100)
```

## 3. Remove Punctuation

Menghapus tanda baca seperti titik, koma, tanda seru, tanda tanya, dan tanda baca lainnya.

```python
def remove_punctuation(text):
    return text.translate(str.maketrans('', '', string.punctuation))

df["Text_Clean"] = df["Text_Clean"].apply(remove_punctuation)
df[["Text", "Text_Clean"]].head(100)
```

## 4. Remove Numbers

Menghapus angka yang terdapat dalam teks.

```python
def remove_numbers(text):
    return re.sub(r'\d+', '', str(text))

df["Text_Clean"] = df["Text_Clean"].apply(remove_numbers)
df[["Text", "Text_Clean"]].head(100)
```

## 5. Remove Extra Spaces

Menghapus spasi berlebih sehingga teks menjadi lebih rapi.

```python
def remove_extra_spaces(text):
    return re.sub(r'\s+', ' ', text).strip()

df["Text_Clean"] = df["Text_Clean"].apply(remove_extra_spaces)
df[["Text", "Text_Clean"]].head(100)
```

menghapus emoji pada judul maupun deskripsi buku jika ada

## 7. Remove Stopwords

Menghapus kata-kata umum dalam bahasa Inggris yang kurang memberikan informasi penting terhadap isi teks.

```python
from nltk.corpus import stopwords

# Mendefinisikan stop_words dari nltk
stop_words = set(stopwords.words('english'))

def remove_stopwords(text):
    words = text.split()
    filtered_words = [word for word in words if word not in stop_words]
    return " ".join(filtered_words)

df["Text_Clean"] = df["Text_Clean"].apply(remove_stopwords)
```

## 7. Stemming

Mengubah kata menjadi bentuk dasar atau bentuk stem dengan menggunakan Porter Stemmer.

```python
from nltk.stem import PorterStemmer

# Mendefinisikan objek stemmer dari PorterStemmer
stemmer = PorterStemmer()

def stemming(text):
    words = text.split()
    stemmed_words = [stemmer.stem(word) for word in words]
    return " ".join(stemmed_words)

df["Text_Clean"] = df["Text_Clean"].apply(stemming)
display(df[["Text", "Text_Clean"]].head())
```

## 8. Removal of Rare words

menghapus kata yang terlalu sering muncul dalam dataset karena dianggap kurang informatif.

```python
from collections import Counter

# Hitung frekuensi setiap kata
all_words = ' '.join(df["Text_Clean"]).split()
word_freq = Counter(all_words)

# Kata yang hanya muncul 1 kali
rare_words = {word for word, freq in word_freq.items() if freq == 1}

print("Jumlah rare words:", len(rare_words))
```

```python
rare_words = {word for word, freq in word_freq.items() if freq <= 2}

print("Jumlah rare words:", len(rare_words))
```

```python
def remove_rare_words(text):
    return " ".join(
        word for word in text.split()
        if word not in rare_words)

df["Text_Clean"] = df["Text_Clean"].apply(remove_rare_words)
df[["Text", "Text_Clean"]].head(10)
```

```python
from gensim.models import Word2Vec

# Tokenisasi
sentences = df["Text_Clean"].apply(lambda x: x.split()).tolist()

# Word2Vec
model_w2v = Word2Vec(
    sentences=sentences,
    vector_size=100,
    window=5,
    min_count=1,
    workers=4
)

# Melihat vektor salah satu kata
word = sentences[0][0]

print("Kata:", word)
print("Vektor:", model_w2v.wv[word])
```

```python
# Untuk mendapatkan fitur berupa satu vektor untuk setiap buku:
import numpy as np

def document_vector(model, words):
    vectors = [model.wv[word] for word in words if word in model.wv]
    if len(vectors) == 0:
        return np.zeros(model.vector_size)
    return np.mean(vectors, axis=0)

w2v_features = np.array([
    document_vector(model_w2v, words)
    for words in sentences
])

print("Ukuran fitur Word2Vec:", w2v_features.shape)
```

## FastText

```python
from gensim.models import FastText

# FastText
model_ft = FastText(
    sentences=sentences,
    vector_size=100,
    window=5,
    min_count=1,
    workers=4
)

word = sentences[0][0]

print("Kata:", word)
print("Vektor:", model_ft.wv[word])
```

```python
# Fitur setiap buku
fasttext_features = np.array([
    document_vector(model_ft, words)
    for words in sentences
])

print("Ukuran fitur FastText:", fasttext_features.shape)
```

# Chapter 8 - Text Clustering

## Elbow Method

```python
# ELBOW METHOD

import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.feature_extraction.text import TfidfVectorizer

# Memastikan X dan tfidf terdefinisi secara global
try:
    X = tfidf
except NameError:
    print("Variabel 'tfidf' tidak ditemukan di memory. Melakukan kalkulasi TF-IDF secara otomatis...")
    tfidf_vectorizer = TfidfVectorizer()
    tfidf = tfidf_vectorizer.fit_transform(df["Text_Clean"])
    X = tfidf

inertia = []
k_values = range(1, 7)

for k in k_values:
    kmeans = KMeans(
        n_clusters=k,
        random_state=10,
        n_init=10)
    kmeans.fit(X)
    inertia.append(kmeans.inertia_)

plt.figure(figsize=(8, 5))
plt.plot(list(k_values), inertia, marker="o")
plt.title("Elbow Method untuk Menentukan Jumlah Cluster")
plt.xlabel("Jumlah Cluster (k)")
plt.ylabel("Inertia")
plt.xticks(list(k_values))
plt.grid(True)
plt.show()
```

## K-Means Clustering

```python
# K-MEANS CLUSTERING

from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
import numpy as np

kmeans = KMeans(
    n_clusters=4,
    random_state=10,
    n_init=10)

# Menggunakan variabel matriks X yang sudah didefinisikan sebelumnya
df["KMeans_Cluster"] = kmeans.fit_predict(X)

# Menghitung Silhouette Score
silhouette = silhouette_score(
    X,
    df["KMeans_Cluster"],
    sample_size=min(500, X.shape[0]),
    random_state=10)

print("Silhouette Score K-Means:", silhouette)
print("\nJumlah data pada setiap cluster:")
print(df["KMeans_Cluster"].value_counts().sort_index())

# Menyesuaikan nama kolom hasil preprocessing terbaru ('Text_Clean')
display(df[["Text_Clean", "KMeans_Cluster"]].head(10))
```

## Agglomerative Clustering

```python
# AGGLOMERATIVE CLUSTERING

from sklearn.decomposition import TruncatedSVD
from sklearn.cluster import AgglomerativeClustering
from sklearn.metrics import silhouette_score

# Mengurangi dimensi fitur TF-IDF
svd = TruncatedSVD(n_components=100, random_state=10)
X_svd = svd.fit_transform(X)

# Membentuk cluster
agg = AgglomerativeClustering(n_clusters=4)
df["Agglomerative_Cluster"] = agg.fit_predict(X_svd)

# Menghitung Silhouette Score
silhouette_agg = silhouette_score(
    X_svd,
    df["Agglomerative_Cluster"],
    sample_size=min(500, X_svd.shape[0]),
    random_state=10)

print("Silhouette Score Agglomerative:", silhouette_agg)
print("\nJumlah data pada setiap cluster:")
print(df["Agglomerative_Cluster"].value_counts().sort_index())

# Menampilkan kolom yang benar-benar ada ('Text_Clean')
display(
    df[["Text_Clean", "Agglomerative_Cluster"]].head(10))
```

## Perbandingan Hasil

```python
# PERBANDINGAN HASIL CLUSTERING

hasil_clustering = pd.DataFrame({
    "Metode": ["K-Means", "Agglomerative"],
    "Silhouette Score": [silhouette, silhouette_agg]})

display(hasil_clustering)

plt.figure(figsize=(7, 4))
plt.bar(
    hasil_clustering["Metode"],
    hasil_clustering["Silhouette Score"])
plt.title("Perbandingan Silhouette Score")
plt.xlabel("Metode Clustering")
plt.ylabel("Silhouette Score")
plt.show()
```

##Evaluasi Silhouette Score


```python
from sklearn.metrics import silhouette_score

# Evaluasi Silhouette Score
print("EVALUASI SILHOUETTE SCORE")
print("=" * 40)

# K-Means
score_kmeans = silhouette_score(X, df["KMeans_Cluster"])
print(f"K-Means Silhouette Score: {score_kmeans:.4f}")

# Agglomerative Clustering
score_agglomerative = silhouette_score(
    X, df["Agglomerative_Cluster"]
)
print(f"Agglomerative Silhouette Score: {score_agglomerative:.4f}")

# Hierarchical Clustering
score_hierarchical = silhouette_score(
    X, df["Hierarchical_Cluster"]
)
print(f"Hierarchical Silhouette Score: {score_hierarchical:.4f}")
```
